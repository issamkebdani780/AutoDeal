## D. Data Visualization

For each plot, add a **title**, **axis labels**, and **measurement units** where applicable. Include a **legend** when colors or shapes identify species. Display the figure with `plt.show()`.

Write one or two sentences of interpretation below each figure. The hints identify useful functions; you must complete the plotting code.


### Task 19. Bar chart: class distribution

Count observations for each species and represent the counts with a bar chart. Determine whether the classes are balanced.

**Guideline:** Count with `value_counts()`. Optional: order the labels using `sort_index()`. Plot the counts with `plot(kind="bar")`.


```python
import matplotlib.pyplot as plt

species_counts = iris_df['species'].value_counts().sort_index()

species_counts.plot(kind='bar', color='skyblue', edgecolor='black')

plt.title('Class Distribution of Iris Species')
plt.xlabel('Species')
plt.ylabel('Number of Observations')

plt.show()
```

### Task 20. Histogram: sepal length

Plot the distribution of sepal length using ten bins. Explain both axes and the height of each bar. Identify the most populated interval.

**Guideline:** Use `plt.hist(...)` with `bins=10`. Each bin is a length interval, and its height counts the flowers in it.


```python
import matplotlib.pyplot as plt

plt.hist(iris_df['sepal_length'], bins=10, edgecolor='black')

plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Number of Flowers')

plt.show()
```

### Task 21. Scatter plot: two measurements

Plot petal length against petal width. Use different colors and marker shapes for the species. Identify the clearly separated species and the species that overlap.

**Guideline:** Use `sns.scatterplot(...)` with `data`, `x`, `y`, `hue` and `style`. Here, x is petal length and y is petal width.


```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.scatterplot(
    data=iris_df,
    x='petal_length',
    y='petal_width',
    hue='species',
    style='species',
    s=80
)

plt.title('Petal Length vs Petal Width by Species')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')

# Display the legend and plot
plt.legend(title='Species')
plt.show()
```

### Task 22. 3D scatter plot

Create a 3D scatter plot with sepal length on x, petal length on y and petal width on z. Distinguish species with colors and a legend. Compare it with the 2D scatter plot and discuss the influence of the viewing angle.

**Guideline:** Create a figure with `plt.figure()`, then an axis with `fig.add_subplot(projection="3d")`. For each species group, call `ax.scatter(x, y, z, label=...)`. Use `ax.set_xlabel(...)`, `ax.set_ylabel(...)` and `ax.set_zlabel(...)`.


```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(projection="3d")

for species in iris_df['species'].unique():
    data = iris_df[iris_df['species'] == species]

    ax.scatter(
        data['sepal_length'],
        data['petal_length'],
        data['petal_width'],
        label=species
    )

ax.set_title('3D Scatter Plot of Iris Species')
ax.set_xlabel('Sepal Length (cm)')
ax.set_ylabel('Petal Length (cm)')
ax.set_zlabel('Petal Width (cm)')

ax.legend(title='Species')

plt.show()
```
